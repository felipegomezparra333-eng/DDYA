### 5) Análisis de complejidad **[15 pts]**
Incluye:
Aspecto	Complejidad
Mejor caso	O(1)

Peor caso	O(n)

Caso promedio	O(n)

Complejidad espacial	O(n)

Fibonacci sin memoización	O(2ⁿ) aproximadamente

Fibonacci con memoización	O(n)

 Si se desea conocer el crecimiento hasta el mes n, el programa debe calcular aproximadamente todos los valores desde F(0) hasta F(n), por lo que la complejidad temporal es O(n). Además, como los resultados calculados se almacenan en memoria para evitar volver a calcularlos, se requiere espacio proporcional al número de meses, dando una complejidad espacial O(n). Gracias a la memoización, esta solución es mucho más eficiente que Fibonacci recursivo sin optimización, cuya complejidad sería aproximadamente O(2ⁿ).

Mejor caso

El mejor caso es O(1) cuando el usuario solicita calcular un mes que corresponde directamente a uno de los casos base, por ejemplo el mes 0, 1 o 2, ya que el programa puede devolver inmediatamente 0, 1 o 1 sin realizar cálculos adicionales. También puede considerarse O(1) si el valor solicitado ya se encuentra almacenado previamente en la memoria de la memoización.

Peor caso

El peor caso es O(n) cuando el usuario solicita un mes n que todavía no ha sido calculado. En ese caso, el programa debe calcular y almacenar progresivamente los valores de Fibonacci desde los casos iniciales hasta llegar al mes solicitado. Aun así, gracias a la memoización, cada resultado se calcula solamente una vez, por lo que el crecimiento del tiempo de ejecución es lineal y no exponencial.

## 6) Código funcional **[15 pts]**
Incluye tu implementación completa.

No pegues código incompleto; debe ser ejecutable.


```python


memo = {}

def sus(m):
    if m < 0:
        return 0

    if m in memo:
        return memo[m]

    if m == 0:
        memo[0] = 0
        return 0

    if m == 1:
        memo[1] = 1
        return 1

    memo[m] = sus(m - 1) + sus(m - 2)
    return memo[m]


def main():
    print("Para acceder diga 'si' o para acabar el programa diga 'no'")
    r = input()

    while r != "no":
        print("¿Desea continuar?")
        r = input()

        if r == "si":
            print("Diga el número de meses a calcular:")
            m = int(input())

            while m < 0 or m >= 1000:
                print("Valor inválido. Introduzca un entero entre 0 y 999:")
                m = int(input())

            resultado = sus(m)
            print("Número de suscriptores:", resultado, "mes:", m)

    print("Gracias")


main()


```
